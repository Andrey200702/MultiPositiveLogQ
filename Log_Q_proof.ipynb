# $L_1$

$$
L_1(u)
=
-\frac{1}{|P(u)|}
\sum_{p\in P(u)}
\log
\frac{e^{s_\theta(u,p)}}
{\sum_{d\in D}e^{s_\theta(u,d)}}
$$

$$
L_1(u)
=
-\frac{1}{m}
\sum_{p\in P}
\log
\frac{e^{s_p}}{Z(u)}
$$

$$
\begin{aligned}
L_1(u)
&=
-\frac{1}{m}
\sum_{p\in P}
\left(
s_p-\log Z(u)
\right)
\\
&=
-\frac{1}{m}
\sum_{p\in P}
s_p
+
\log Z(u)
\end{aligned}
$$

# $\nabla_\theta L_1$

$$
\begin{aligned}
\nabla_\theta L_1(u)
&=
-\frac{1}{m}
\sum_{p\in P}
\nabla_\theta s_p
+
\nabla_\theta \log Z(u)
\\
&=
-\frac{1}{m}
\sum_{p\in P}
\nabla_\theta s_p
+
\frac{1}{Z(u)}
\sum_{d\in D}
e^{s_d}
\nabla_\theta s_d
\\
&=
-\frac{1}{m}
\sum_{p\in P}
\nabla_\theta s_p
+
\sum_{d\in D}
\pi_\theta(d\mid u)
\nabla_\theta s_d
\end{aligned}
$$

$$
\boxed{
\nabla_\theta L_1(u)
=
\sum_{d\in D}
\pi_\theta(d\mid u)
\nabla_\theta s_\theta(u,d)
-
\frac{1}{|P(u)|}
\sum_{p\in P(u)}
\nabla_\theta s_\theta(u,p)
}
$$

# $L_2$

$$
L_2(u)
=
-\log
\left(
\frac{1}{|P(u)|}
\sum_{p\in P(u)}
\frac{e^{s_\theta(u,p)}}
{\sum_{d\in D}e^{s_\theta(u,d)}}
\right)
$$

$$
L_2(u)
=
-\log
\left(
\frac{1}{m}
\sum_{p\in P}
\frac{e^{s_p}}{Z(u)}
\right)
$$

$$
\begin{aligned}
L_2(u)
&=
-\log
\left(
\frac{1}{m}
\frac{A_P(u)}{Z(u)}
\right)
\\
&=
-\log A_P(u)
+
\log Z(u)
+
\log m
\end{aligned}
$$

# $\nabla_\theta L_2$

$$
\begin{aligned}
\nabla_\theta L_2(u)
&=
-\nabla_\theta \log A_P(u)
+
\nabla_\theta \log Z(u)
\\
&=
-
\frac{1}{A_P(u)}
\sum_{p\in P}
e^{s_p}
\nabla_\theta s_p
+
\frac{1}{Z(u)}
\sum_{d\in D}
e^{s_d}
\nabla_\theta s_d
\\
&=
-
\sum_{p\in P}
\pi_\theta^+(p\mid u)
\nabla_\theta s_p
+
\sum_{d\in D}
\pi_\theta(d\mid u)
\nabla_\theta s_d
\end{aligned}
$$

$$
\boxed{
\nabla_\theta L_2(u)
=
\sum_{d\in D}
\pi_\theta(d\mid u)
\nabla_\theta s_\theta(u,d)
-
\sum_{p\in P(u)}
\pi_\theta^+(p\mid u)
\nabla_\theta s_\theta(u,p)
}
$$

# Log-$q$

$$
C(u)=P(u)\cup N(u)
$$

$$
q(c\mid u),
\qquad
c\in C(u)
$$

$$
\widetilde{s}_c
=
s_c-\log q(c\mid u)
$$

$$
e^{\widetilde{s}_c}
=
e^{s_c-\log q(c\mid u)}
=
\frac{e^{s_c}}{q(c\mid u)}
$$

$$
q(p\mid u)=1,
\qquad
\widetilde{s}_p=s_p,
\qquad
p\in P(u)
$$

$$
\widetilde{Z}(u)
=
\sum_{c\in C(u)}e^{\widetilde{s}_c}
=
\sum_{c\in C(u)}
e^{s_c-\log q(c\mid u)}
$$

$$
\widetilde{\pi}_\theta(c\mid u)
=
\frac{e^{\widetilde{s}_c}}{\widetilde{Z}(u)}
=
\frac{e^{s_c-\log q(c\mid u)}}
{\sum_{c'\in C(u)}e^{s_{c'}-\log q(c'\mid u)}}
$$

$$
\nabla_\theta \widetilde{s}_c
=
\nabla_\theta s_c
$$

# $\widetilde{L}_1$

$$
\widetilde{L}_1(u)
=
-\frac{1}{|P(u)|}
\sum_{p\in P(u)}
\log
\frac{e^{\widetilde{s}_p}}
{\sum_{c\in C(u)}e^{\widetilde{s}_c}}
$$

$$
\widetilde{L}_1(u)
=
-\frac{1}{m}
\sum_{p\in P}
\log
\frac{e^{\widetilde{s}_p}}
{\widetilde{Z}(u)}
$$

$$
\begin{aligned}
\widetilde{L}_1(u)
&=
-\frac{1}{m}
\sum_{p\in P}
\left(
\widetilde{s}_p-\log \widetilde{Z}(u)
\right)
\\
&=
-\frac{1}{m}
\sum_{p\in P}
\widetilde{s}_p
+
\log \widetilde{Z}(u)
\end{aligned}
$$

# $\nabla_\theta \widetilde{L}_1$

$$
\begin{aligned}
\nabla_\theta \widetilde{L}_1(u)
&=
-\frac{1}{m}
\sum_{p\in P}
\nabla_\theta \widetilde{s}_p
+
\nabla_\theta \log \widetilde{Z}(u)
\\
&=
-\frac{1}{m}
\sum_{p\in P}
\nabla_\theta s_p
+
\frac{1}{\widetilde{Z}(u)}
\sum_{c\in C(u)}
e^{\widetilde{s}_c}
\nabla_\theta \widetilde{s}_c
\\
&=
-\frac{1}{m}
\sum_{p\in P}
\nabla_\theta s_p
+
\frac{1}{\widetilde{Z}(u)}
\sum_{c\in C(u)}
e^{\widetilde{s}_c}
\nabla_\theta s_c
\\
&=
-\frac{1}{m}
\sum_{p\in P}
\nabla_\theta s_p
+
\sum_{c\in C(u)}
\widetilde{\pi}_\theta(c\mid u)
\nabla_\theta s_c
\end{aligned}
$$

$$
\boxed{
\nabla_\theta \widetilde{L}_1(u)
=
\sum_{c\in C(u)}
\widetilde{\pi}_\theta(c\mid u)
\nabla_\theta s_\theta(u,c)
-
\frac{1}{|P(u)|}
\sum_{p\in P(u)}
\nabla_\theta s_\theta(u,p)
}
$$

# $\widetilde{L}_2$

$$
\widetilde{L}_2(u)
=
-\log
\left(
\frac{1}{|P(u)|}
\sum_{p\in P(u)}
\frac{e^{\widetilde{s}_p}}
{\sum_{c\in C(u)}e^{\widetilde{s}_c}}
\right)
$$

$$
\widetilde{L}_2(u)
=
-\log
\left(
\frac{1}{m}
\sum_{p\in P}
\frac{e^{\widetilde{s}_p}}
{\widetilde{Z}(u)}
\right)
$$

$$
\widetilde{A}_P(u)
=
\sum_{p\in P(u)}e^{\widetilde{s}_p}
=
\sum_{p\in P(u)}
e^{s_p-\log q(p\mid u)}
$$

$$
\begin{aligned}
\widetilde{L}_2(u)
&=
-\log
\left(
\frac{1}{m}
\frac{\widetilde{A}_P(u)}
{\widetilde{Z}(u)}
\right)
\\
&=
-\log \widetilde{A}_P(u)
+
\log \widetilde{Z}(u)
+
\log m
\end{aligned}
$$

$$
\widetilde{\pi}_\theta^+(p\mid u)
=
\frac{e^{\widetilde{s}_p}}
{\widetilde{A}_P(u)}
=
\frac{e^{s_p-\log q(p\mid u)}}
{\sum_{p'\in P(u)}e^{s_{p'}-\log q(p'\mid u)}}
$$

$$
\begin{aligned}
\nabla_\theta \widetilde{L}_2(u)

-\nabla_\theta \log \widetilde{A}_P(u)

\nabla_\theta \log \widetilde{Z}(u)

-
\frac{1}{\widetilde{A}_P(u)}
\sum_{p\in P}
e^{\widetilde{s}_p}
\nabla_\theta \widetilde{s}_p

\frac{1}{\widetilde{Z}(u)}
\sum_{c\in C(u)}
e^{\widetilde{s}_c}
\nabla_\theta \widetilde{s}_c

-
\frac{1}{\widetilde{A}_P(u)}
\sum_{p\in P}
e^{\widetilde{s}_p}
\nabla_\theta s_p
+
\frac{1}{\widetilde{Z}(u)}
\sum_{c\in C(u)}
e^{\widetilde{s}_c}
\nabla_\theta s_c
\\
&=
-
\sum_{p\in P}
\widetilde{\pi}_\theta^+(p\mid u)
\nabla_\theta s_p
+
\sum_{c\in C(u)}
\widetilde{\pi}_\theta(c\mid u)
\nabla_\theta s_c
\end{aligned}
$$

$$
\boxed{
\nabla_\theta \widetilde{L}_2(u)
=
\sum_{c\in C(u)}
\widetilde{\pi}_\theta(c\mid u)
